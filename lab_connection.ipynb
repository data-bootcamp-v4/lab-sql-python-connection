```python
import pandas as pd
import numpy as np
import pymysql
from sqlalchemy import create_engine
from sqlalchemy import text
from sqlalchemy.engine import URL
```

```python
bd = "sakila"
connection_string = 'mysql+pymysql://root:' + 'fuTsC4Os5mrtE59' + '@localhost/'+bd
engine = create_engine(connection_string)
engine
```

```python
def rentals_month(engine, month, year):
    with engine.connect() as connection:
        query = text('SELECT * FROM rental WHERE YEAR(rental_date) = :year AND MONTH(rental_date) = :month')
        result = connection.execute(query, {"year": year, "month": month})
        rentals = pd.DataFrame(result.all())
    return rentals
```

```python
rentals = rentals_month(engine,6,2005)
```

```python
def rental_count_month(rentals, year, month):
    rentals[f'rentals_{month}_{year}'] = rentals.groupby('customer_id')['rental_id'].count()
    rentals_month = rentals[["customer_id",f'rentals_{month}_{year}']]
    return rentals_month
```
