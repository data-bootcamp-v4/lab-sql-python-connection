```python
import pandas as pd
import numpy as np
import pymysql
from sqlalchemy import create_engine, text # TEX HAS TO BE IMPORT TOO
import getpass  # To get the password without showing the input
password = getpass.getpass()
```

```python
sd = "sakila"
connection_string = 'mysql+pymysql://root:' + password + '@localhost/'+sd
engine = create_engine(connection_string)
engine
```

```python
def rentals_month(engine, month, year):
    with engine.connect() as connection:
        txt= f'''SELECT *
                FROM rental
                WHERE month(rental_date) = {month} and year(rental_date) = {year};'''
        query = text(txt)
        result = connection.execute(query)
        return pd.DataFrame(result.all())
```

```python
rentals_month(engine, 5,2005)
```

```python
#Develop a Python function called rental_count_month that takes the DataFrame 
#provided by rentals_month as input along with the month and year and returns a new DataFrame 
#containing the number of rentals made by each customer_id during the selected month and year.
```

```python
def rental_month_count(rentals_month,m,y):
    df= rentals_month.groupby("customer_id").agg("count")
    df=df.rename(columns={"rental_date":f"rentals_{m}_{y}"})
    df=df.drop(columns=["rental_id","inventory_id","staff_id","last_update","return_date"])
    return df

rental_month_count(rentals_month(engine,5,2005),5,2005)
```

```python
def compare_rentals(df1,df2):
    df=pd.merge(df1,df2, on="customer_id")
    df["difference"]=df[df.columns[0]]-df[df.columns[1]]
    return df
```
