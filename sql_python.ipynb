```python
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:11223344@localhost/sakila")
```

```python
import pandas as pd

def rentals_month(engine, month, year):
    """
    Fetch all rental records for a given month and year.
    """

    query = f"""
        SELECT *
        FROM rental
        WHERE MONTH(rental_date) = {month}
        AND YEAR(rental_date) = {year};
    """

    with engine.connect() as connection:
        df = pd.read_sql_query(query, connection)

    return df
```

```python
def rental_count_month(df, month, year):
    """
    Count how many rentals each customer made in a month.
    """

    column_name = f"rentals_{str(month).zfill(2)}_{year}"

    result = (
        df.groupby("customer_id")
          .size()
          .reset_index(name=column_name)
    )

    return result
```

```python
def compare_rentals(df1, df2):
    """
    Compare rental counts between two months.
    """

    merged = pd.merge(df1, df2, on="customer_id", how="outer").fillna(0)

    # Get the two rental columns (excluding customer_id)
    cols = [col for col in merged.columns if col != "customer_id"]

    merged["difference"] = merged[cols[1]] - merged[cols[0]]

    return merged
```

```python
# Step 1: Load data
may_data = rentals_month(engine, 5, 2005)
june_data = rentals_month(engine, 6, 2005)

# Step 2: Count rentals
may_counts = rental_count_month(may_data, 5, 2005)
june_counts = rental_count_month(june_data, 6, 2005)

# Step 3: Compare
comparison = compare_rentals(may_counts, june_counts)

# Step 4: View result
print(comparison.head())
```

```python
import pandas as pd
from sqlalchemy import text

def search_customers(engine, keyword):

    query = text("""
        SELECT 
            customer_id,
            first_name,
            last_name,
            email
        FROM customer
        WHERE first_name LIKE :kw
           OR last_name LIKE :kw;
    """)

    params = {"kw": f"%{keyword}%"}

    with engine.connect() as conn:
        df = pd.read_sql(query, conn, params=params)

    return df
```

```python
keyword = input("Enter name: ")

df = search_customers(engine, keyword)

df
df
```
