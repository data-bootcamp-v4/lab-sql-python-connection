```python
from sqlalchemy import create_engine
import pandas as pd

db_user = 'root'
db_password = 'Burningcaravan123'
db_host = 'localhost'
db_port = '3306'
db_name = 'sakila'

engine = create_engine(f"mysql+pymysql://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}")
```

```python
#crear funcion
def rentals_month(engine, month, year):
    query=f"""
    SELECT rental_id, customer_id, rental_date
    FROM rental
    WHERE MONTH(rental_date)= {month} AND YEAR(rental_date)={year};
    """
    return pd.read_sql(query, con=engine)
```

```python
#funcion rental_count_month

def rental_count_month(df, month, year):
    rental_col=f"rentals_{month:02d}_{year}"
    rental_count_df=df.groupby('customer_id').size().reset_index(name=rental_col)
    return rental_count_df
```

```python
#funcion compare_rentals

def compare_rentals(df1,df2):
    df_combined=pd.merge(df1,df2, on='customer_id',how='inner')
    col1= df1.columns[1]
    col2= df2.columns[1]
    df_combined['difference']= df_combined[col2]-df_combined[col1]
    return df_combined
```

```python
#ej

# P1: obtener datos de mayo y junio
df_may = rentals_month(engine, 5, 2005)
df_june = rentals_month(engine, 6, 2005)

# P2: contar alquileres por cliente
may_rentals = rental_count_month(df_may, 5, 2005)
june_rentals = rental_count_month(df_june, 6, 2005)

# P3: comparar alquileres
comparison_df = compare_rentals(may_rentals, june_rentals)

print(comparison_df.head())
```
