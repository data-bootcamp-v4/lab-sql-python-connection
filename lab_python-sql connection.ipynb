```python
import pandas as pd
from sqlalchemy import create_engine

# conexión
engine = create_engine("mysql+pymysql://furby:1234@localhost/sakila")

# 1. rentals_month
def rentals_month(engine, month, year):
    query = f"""
    SELECT *
    FROM rental
    WHERE MONTH(rental_date) = {month}
    AND YEAR(rental_date) = {year};
    """
    return pd.read_sql(query, engine)

# 2. rental_count_month
def rental_count_month(df, month, year):
    col_name = f"rentals_{str(month).zfill(2)}_{year}"
    
    return (
        df.groupby("customer_id")
        .size()
        .reset_index(name=col_name)
    )

# 3. compare_rentals
def compare_rentals(df1, df2):
    merged = pd.merge(df1, df2, on="customer_id", how="outer").fillna(0)
    
    col1 = df1.columns[1]
    col2 = df2.columns[1]
    
    merged["difference"] = merged[col2] - merged[col1]
    
    return merged

# TEST
may = rentals_month(engine, 5, 2005)
may_count = rental_count_month(may, 5, 2005)

jun = rentals_month(engine, 6, 2005)
jun_count = rental_count_month(jun, 6, 2005)

result = compare_rentals(may_count, jun_count)

print(result.head())
```
