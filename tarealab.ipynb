## ejercicio 2

```python
import pandas as pd

def rental_month(engine,month,year):

    query="SELECT * FROM rental WHERE MONTH(rental_date) = %s AND YEAR(rental_date) = %s"
    df=pd.read_sql(query, engine, params=(month, year))
    return df
```

```python
dfrental=rental_month(engine,6,2005)
dfrental.head()
```

## ejercicio 3

```python

def rental_count_month(df, month, year):
    df_counts = df.groupby('customer_id')[['rental_id']].count().reset_index()
    new_col_name = f"alquileres_{month:02d}_{year}"
    df_counts.columns = ['customer_id', new_col_name]
    
    return df_counts
```

```python
df_rental_count_month=rental_count_month(dfrental,5,2005)
```

```python
df_rental_count_month.head()
```

## ejercicio 4

```python
def compare_rentals(df1,df2):
    df_merge=pd.merge(df1,df2,on='customer_id',how='inner')
    
    return df_merge
    
```

```python
df_final=compare_rentals(df_rental_count_month,dfrental)
df_final.head()
```
