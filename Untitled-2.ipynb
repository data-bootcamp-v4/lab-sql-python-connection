```python
# pip install pandas sqlalchemy pymysql
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus
from datetime import date

# ---- 0) DB connection (edit user/pass/host/port/db) ----
USER = "root"
PASSWORD = quote_plus("Fahrscheine1!")   # quote_plus handles special chars
HOST = "127.0.0.1"
PORT = 3306
DB   = "sakila"

engine = create_engine(
    f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB}?charset=utf8mb4",
    future=True
)

# ---- 1) rentals_month ----
def rentals_month(engine, month: int, year: int) -> pd.DataFrame:
    """
    Return all rows from sakila.rental for the given month/year.
    Uses a half-open date range [start, end) for index friendliness.
    """
    if not (1 <= month <= 12):
        raise ValueError("month must be 1..12")

    start = date(year, month, 1)
    # first day of next month
    end = date(year + (month == 12), (month % 12) + 1, 1)

    sql = text("""
        SELECT rental_id, rental_date, inventory_id, customer_id, staff_id, last_update
        FROM rental
        WHERE rental_date >= :start AND rental_date < :end
        ORDER BY rental_date
    """)
    with engine.connect() as conn:
        df = pd.read_sql(sql, conn, params={"start": start, "end": end})
    return df

# ---- 2) rental_count_month ----
def rental_count_month(df: pd.DataFrame, month: int, year: int) -> pd.DataFrame:
    """
    Count rentals per customer_id for the given month's DataFrame.
    Output: columns = ['customer_id', f'rentals_{MM}_{YYYY}']
    """
    col = f"rentals_{month:02d}_{year}"
    return (
        df.groupby("customer_id", as_index=False)
          .agg(**{col: ("rental_id", "count")})
    )

# ---- 3) compare_rentals ----
def compare_rentals(df_first: pd.DataFrame, df_second: pd.DataFrame) -> pd.DataFrame:
    """
    Merge two monthly counts on customer_id (INNER JOIN = active in both).
    Adds 'difference' = (second month) - (first month).
    """
    col_first  = [c for c in df_first.columns  if c != "customer_id"][0]
    col_second = [c for c in df_second.columns if c != "customer_id"][0]

    merged = df_first.merge(df_second, on="customer_id", how="inner")
    merged["difference"] = merged[col_second] - merged[col_first]
    return merged

# ---------------- Example: May vs June 2005 ----------------
if __name__ == "__main__":
    may = rentals_month(engine, 5, 2005)
    jun = rentals_month(engine, 6, 2005)

    may_counts = rental_count_month(may, 5, 2005)    # customer_id, rentals_05_2005
    jun_counts = rental_count_month(jun, 6, 2005)    # customer_id, rentals_06_2005

    both_active = compare_rentals(may_counts, jun_counts)  # only customers active in BOTH
    # Who increased most from May to June?
    print(both_active.sort_values("difference", ascending=False).head(10))
```
